## Objective

Clean and preprocess a 2019 online vehicle dataset (≈15,919 rows, 54 columns, 9 car models) containing many inconsistent, broken, and messy values, so it becomes suitable for machine learning. The project is broken up into four main phases: 

1. Data Cleaning: 
    * Removing broken, irrelevant, or redundant columns
    * Creating new columns with clean, meaningful, and usable values
2. Missing Value Analysis:
    * Filling in the gabs where data is missing
    * With techniques such as mean, median, or mode
3. Outlier Analysis:
    * Outliers are detected and visualized using boxplots and histograms.
    * Rather than removing all outliers (since extreme but valid values are expected for certain car models, e.g. luxury vehicles with high prices or powerful engines), the focus is on identifying and addressing only clear misprints or logically inconsistent entries.
4. Dummy Variable Analysis:
    * Categorical and multi-label columns (e.g. Comfort_Convenience, Extras) are converted to numeric format to create binary indicator variables suitable for machine learning.
    * Key techniques include pd.get_dummies() and str.get_dummies(sep=",")

This notebook covers Phase 4 Dummy Variable Analysis with input from phase3_cleaned_data.csv (15919 rows, 34 columns) and output cleaned_data.csv.

## Libraries

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore", category=SyntaxWarning)
```

## Import Data

```python
df_org = pd.read_csv("data/phase3_cleaned_data.csv")
df = df_org.copy()
df.head(3).T
```

```python
df.info()
```

## Dummy Variable Analysis

```python
exclude = ['comfort_convenience', 'entertainment_media', 'extras', 'safety_security']

dummy_cols = [col for col in df.select_dtypes(include='str') if col not in exclude]

dummy_df = pd.get_dummies(df, columns=dummy_cols, drop_first=True, dtype=int)

dummy_df.shape
```

### comfort_convenience

```python
df['comfort_convenience'].value_counts()
```

```python
dummy_cc = df['comfort_convenience'].str.get_dummies(sep=',').add_prefix(f"comfort_convenience_")

sorted_cc = dummy_cc.sum().sort_values(ascending=True)

sorted_cc
```

```python
dropped_cc = sorted_cc.head(5).index.tolist()

print(f"Dropped: {dropped_cc}")
```

```python
dummy_cc = dummy_cc.drop(columns=dropped_cc)
```

### entertainment_media

```python
df['entertainment_media'].value_counts()
```

```python
dummy_em = df['entertainment_media'].str.get_dummies(sep=',').add_prefix(f"entertainment_media_")

sorted_em = dummy_em.sum().sort_values(ascending=True)

sorted_em
```

```python
dropped_em = sorted_em.head(1).index.tolist()

print(f"Dropped: {dropped_em}")
```

```python
dummy_em = dummy_em.drop(columns=dropped_em)
```

### extras

```python
df['extras'].value_counts()
```

```python
dummy_e = df['extras'].str.get_dummies(sep=',').add_prefix(f"extras_")

sorted_e = dummy_e.sum().sort_values(ascending=True)

sorted_e
```

```python
dropped_e = sorted_e.head(4).index.tolist()

print(f"Dropped: {dropped_e}")
```

```python
dummy_e = dummy_e.drop(columns=dropped_e)
```

### safety_security

```python
df['safety_security'].value_counts()
```

```python
dummy_ss = df['safety_security'].str.get_dummies(sep=',').add_prefix(f"safety_security_")

sorted_ss = dummy_ss.sum().sort_values(ascending=True)

sorted_ss
```

```python
dropped_ss = sorted_ss.head(1).index.tolist()

print(f"Dropped: {dropped_ss}")
```

## Concatenate Dummy Variables

```python
df = pd.concat([dummy_df, dummy_cc, dummy_e, dummy_ss, dummy_em], axis=1)

df = df.drop(columns=exclude)

df.shape
```

## Standardize Columns Names

```python
df.columns.str.replace(' ', '_').str.replace('-', '_').str.lower()
```

```python
df.columns = df.columns.str.replace(' ', '_').str.replace('-', '_').str.lower()
```
